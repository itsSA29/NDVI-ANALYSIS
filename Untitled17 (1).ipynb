```python
# ============================================
# STEP 1: UPLOAD AND LOAD CLEANED DATASET
# ============================================

import pandas as pd
import numpy as np
from google.colab import files

uploaded = files.upload()

filename = list(uploaded.keys())[0]

df = pd.read_excel(filename)

print("Dataset loaded successfully!")
print("----------------------------------------")
print("Rows   :", df.shape[0])
print("Columns:", df.shape[1])
print("Columns:", df.columns.tolist())
```

```python
# ============================================
# STEP 2: CREATE 10,000-ROW DATASET
# ============================================

df_model = df.sample(
    n=10000,
    random_state=42
).reset_index(drop=True)

print("10,000-row dataset created successfully!")
print("--------------------------------------------")
print("Rows   :", df_model.shape[0])
print("Columns:", df_model.shape[1])

print("\nColumns:")
print(df_model.columns.tolist())

print("\nFirst 5 rows:")
display(df_model.head())
```

```python
# ============================================
# STEP 3: PREPARE INPUTS AND OUTPUT
# ============================================

# INPUTS
X = df_model[
    [
        "PRECIPITATION",
        "LST",
        "ETo"
    ]
]

# OUTPUT 1: NDVI value
y = df_model["NDVI"]

print("Inputs and output prepared successfully!")
print("--------------------------------------------")

print("\nINPUTS:")
print(X.columns.tolist())

print("\nNumber of inputs:", X.shape[1])

print("\nOUTPUT:")
print("Predicted NDVI")

print("\nInput shape :", X.shape)
print("Output shape:", y.shape)

print("\nFirst 5 input rows:")
display(X.head())

print("\nFirst 5 NDVI values:")
display(y.head())
```

```python
# ============================================
# STEP 4: TRAIN-TEST SPLIT
# ============================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Train-test split completed successfully!")
print("--------------------------------------------")

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))

print("\nTraining input shape:", X_train.shape)
print("Testing input shape :", X_test.shape)

print("\nTraining output shape:", y_train.shape)
print("Testing output shape :", y_test.shape)
```

```python
# ============================================
# STEP 5: CHECK RELATIONSHIP WITH NDVI
# ============================================

correlation = df_model[
    ["PRECIPITATION", "LST", "ETo", "NDVI"]
].corr()

print("CORRELATION WITH NDVI")
print("======================")

print("\nPrecipitation vs NDVI:",
      round(correlation.loc["PRECIPITATION", "NDVI"], 4))

print("LST vs NDVI:",
      round(correlation.loc["LST", "NDVI"], 4))

print("ETo vs NDVI:",
      round(correlation.loc["ETo", "NDVI"], 4))

print("\nComplete correlation matrix:")
display(correlation)
```

```python
# ============================================
# STEP 5: VISUALIZE RELATIONSHIPS WITH NDVI
# ============================================

import matplotlib.pyplot as plt

# Precipitation vs NDVI
plt.figure(figsize=(7,5))
plt.scatter(df_model["PRECIPITATION"], df_model["NDVI"], alpha=0.5)
plt.xlabel("Precipitation")
plt.ylabel("NDVI")
plt.title("Precipitation vs NDVI")
plt.grid(True)
plt.show()


# LST vs NDVI
plt.figure(figsize=(7,5))
plt.scatter(df_model["LST"], df_model["NDVI"], alpha=0.5)
plt.xlabel("LST")
plt.ylabel("NDVI")
plt.title("LST vs NDVI")
plt.grid(True)
plt.show()


# ETo vs NDVI
plt.figure(figsize=(7,5))
plt.scatter(df_model["ETo"], df_model["NDVI"], alpha=0.5)
plt.xlabel("ETo")
plt.ylabel("NDVI")
plt.title("ETo vs NDVI")
plt.grid(True)
plt.show()
```

```python
# ============================================
# STEP 6: TRAIN-TEST SPLIT
# ============================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Train-test split completed successfully!")
print("--------------------------------------------")
print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))

print("\nTraining input shape:", X_train.shape)
print("Testing input shape :", X_test.shape)

print("\nTraining output shape:", y_train.shape)
print("Testing output shape :", y_test.shape)
```

```python
# ============================================
# STEP 7: RANDOM FOREST NDVI PREDICTION
# ============================================

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

print("Creating Random Forest model...")

# Create Random Forest Regression model
rf_ndvi = RandomForestRegressor(
    n_estimators=100,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)

# Train the model
print("Training Random Forest...")
rf_ndvi.fit(X_train, y_train)

print("Random Forest trained successfully!")

# Predict NDVI for test data
y_pred = rf_ndvi.predict(X_test)

# Calculate performance
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("\n========================================")
print("NDVI PREDICTION RESULTS")
print("========================================")

print("MAE  :", round(mae, 4))
print("RMSE :", round(rmse, 4))
print("R²   :", round(r2, 4))

# Compare actual and predicted NDVI
results = pd.DataFrame({
    "Actual NDVI": y_test.values,
    "Predicted NDVI": y_pred
})

print("\nFirst 10 predictions:")
display(results.head(10))
```

```python
# ============================================
# STEP 8: VEGETATION HEALTH CLASSIFICATION
# ============================================

def vegetation_health(ndvi):
    if ndvi < 0.20:
        return "Stressed"
    elif ndvi < 0.50:
        return "Moderate"
    else:
        return "Healthy"


# Convert predicted NDVI into vegetation health
health_prediction = [vegetation_health(value) for value in y_pred]

# Create final prediction table
final_results = pd.DataFrame({
    "Actual NDVI": y_test.values,
    "Predicted NDVI": y_pred,
    "Vegetation Health": health_prediction
})

print("========================================")
print("VEGETATION HEALTH PREDICTION")
print("========================================")

display(final_results.head(10))

print("\nVegetation Health Distribution:")
print(pd.Series(health_prediction).value_counts())
```

```python
# ============================================
# STEP 9: FINAL VEGETATION HEALTH PREDICTION
# ============================================

print("========================================")
print("VEGETATION HEALTH PREDICTION SYSTEM")
print("========================================")

# Enter the 3 input values used by the trained model
precipitation = float(input("Enter Precipitation: "))
lst = float(input("Enter LST: "))
eto = float(input("Enter ETo: "))

# Create input using EXACTLY the features used during training
new_input = pd.DataFrame({
    "PRECIPITATION": [precipitation],
    "LST": [lst],
    "ETo": [eto]
})

# Predict NDVI
predicted_ndvi = rf_ndvi.predict(new_input)[0]

# Determine vegetation health from predicted NDVI
health = vegetation_health(predicted_ndvi)

# Display final result
print("\n========================================")
print("FINAL PREDICTION")
print("========================================")

print("Precipitation       :", precipitation)
print("LST                 :", lst)
print("ETo                 :", eto)

print("\nPredicted NDVI      :", round(predicted_ndvi, 4))
print("Vegetation Health   :", health)

print("========================================")
```

```python
# ============================================
# STEP 10: ACTUAL vs PREDICTED NDVI
# ============================================

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.5
)

# Perfect prediction reference line
min_value = min(y_test.min(), y_pred.min())
max_value = max(y_test.max(), y_pred.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.xlabel("Actual NDVI")
plt.ylabel("Predicted NDVI")
plt.title("Actual vs Predicted NDVI")

plt.grid(True)
plt.show()
```

```python
# ============================================
# STEP 11: FEATURE IMPORTANCE
# ============================================

import matplotlib.pyplot as plt

# Get feature importance from the trained Random Forest
importance = rf_ndvi.feature_importances_

features = ["Precipitation", "LST", "ETo"]

print("========================================")
print("FEATURE IMPORTANCE FOR NDVI")
print("========================================")

for feature, value in zip(features, importance):
    print(f"{feature}: {value:.4f}")

# Plot
plt.figure(figsize=(8, 5))
plt.bar(features, importance)

plt.xlabel("Environmental Factors")
plt.ylabel("Importance")
plt.title("Feature Importance for NDVI Prediction")
plt.grid(axis="y")

plt.show()
```

```python
# ============================================
# STEP 12: FINAL MODEL EVALUATION
# ============================================

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("========================================")
print("FINAL RANDOM FOREST MODEL EVALUATION")
print("========================================")

print("Mean Absolute Error (MAE) :", round(mae, 4))
print("Root Mean Squared Error (RMSE) :", round(rmse, 4))
print("R² Score :", round(r2, 4))

print("\nR² Score in percentage :", round(r2 * 100, 2), "%")

print("========================================")
```

```python
# ============================================
# STEP 13: VEGETATION HEALTH EVALUATION
# ============================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Convert NDVI values into health classes
actual_health = [
    vegetation_health(value) for value in y_test
]

predicted_health = [
    vegetation_health(value) for value in y_pred
]

# Calculate classification metrics
accuracy = accuracy_score(actual_health, predicted_health)

precision = precision_score(
    actual_health,
    predicted_health,
    average="macro",
    zero_division=0
)

recall = recall_score(
    actual_health,
    predicted_health,
    average="macro",
    zero_division=0
)

f1 = f1_score(
    actual_health,
    predicted_health,
    average="macro",
    zero_division=0
)

print("========================================")
print("VEGETATION HEALTH EVALUATION")
print("========================================")

print("Accuracy :", round(accuracy, 4))
print("Accuracy %:", round(accuracy * 100, 2), "%")

print("Macro Precision:", round(precision, 4))
print("Macro Recall   :", round(recall, 4))
print("Macro F1-score :", round(f1, 4))

print("\n========================================")
print("CLASSIFICATION REPORT")
print("========================================")

print(
    classification_report(
        actual_health,
        predicted_health,
        labels=["Stressed", "Moderate", "Healthy"],
        zero_division=0
    )
)
```

```python
# ============================================
# STEP 14: CONFUSION MATRIX
# ============================================

from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt

# Define class order
classes = ["Stressed", "Moderate", "Healthy"]

# Create confusion matrix
cm = confusion_matrix(
    actual_health,
    predicted_health,
    labels=classes
)

print("========================================")
print("CONFUSION MATRIX")
print("========================================")

print(cm)

# Plot confusion matrix
plt.figure(figsize=(7, 6))

plt.imshow(cm)

plt.xticks(range(len(classes)), classes)
plt.yticks(range(len(classes)), classes)

plt.xlabel("Predicted Health")
plt.ylabel("Actual Health")
plt.title("Vegetation Health Confusion Matrix")

# Add numbers inside the matrix
for i in range(len(classes)):
    for j in range(len(classes)):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.colorbar()
plt.show()
```

```python
# ============================================
# STEP 15: RANDOM FOREST HYPERPARAMETER TUNING
# ============================================

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

print("========================================")
print("TUNING RANDOM FOREST FOR NDVI")
print("========================================")

# Different model settings to test
settings = [
    {"n_estimators": 100, "max_depth": 10},
    {"n_estimators": 200, "max_depth": 15},
    {"n_estimators": 300, "max_depth": 20},
    {"n_estimators": 300, "max_depth": None}
]

best_model = None
best_r2 = -999

for setting in settings:

    model = RandomForestRegressor(
        n_estimators=setting["n_estimators"],
        max_depth=setting["max_depth"],
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_train, y_train)

    prediction = model.predict(X_test)

    mae = mean_absolute_error(y_test, prediction)
    rmse = np.sqrt(mean_squared_error(y_test, prediction))
    r2 = r2_score(y_test, prediction)

    print("\n----------------------------------------")
    print("Trees:", setting["n_estimators"])
    print("Max Depth:", setting["max_depth"])
    print("MAE :", round(mae, 4))
    print("RMSE:", round(rmse, 4))
    print("R²  :", round(r2, 4))

    if r2 > best_r2:
        best_r2 = r2
        best_model = model

print("\n========================================")
print("BEST MODEL SELECTED")
print("========================================")
print("Best R²:", round(best_r2, 4))
print("Best model:", best_model)
```

```python
# ============================================
# STEP 16: EVALUATE TUNED MODEL
# ============================================

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Create the best tuned Random Forest again
tuned_model = RandomForestRegressor(
    n_estimators=300,
    max_depth=None,
    random_state=42,
    n_jobs=-1
)

# Train the tuned model
tuned_model.fit(X_train, y_train)

# Predict NDVI
tuned_pred = tuned_model.predict(X_test)

# Convert NDVI to vegetation health classes
tuned_actual_health = [
    vegetation_health(value) for value in y_test
]

tuned_predicted_health = [
    vegetation_health(value) for value in tuned_pred
]

# Calculate classification metrics
tuned_accuracy = accuracy_score(
    tuned_actual_health,
    tuned_predicted_health
)

tuned_precision = precision_score(
    tuned_actual_health,
    tuned_predicted_health,
    average="macro",
    zero_division=0
)

tuned_recall = recall_score(
    tuned_actual_health,
    tuned_predicted_health,
    average="macro",
    zero_division=0
)

tuned_f1 = f1_score(
    tuned_actual_health,
    tuned_predicted_health,
    average="macro",
    zero_division=0
)

print("========================================")
print("TUNED MODEL - VEGETATION HEALTH")
print("========================================")

print("Accuracy :", round(tuned_accuracy, 4))
print("Accuracy %:", round(tuned_accuracy * 100, 2), "%")
print("Macro Precision:", round(tuned_precision, 4))
print("Macro Recall   :", round(tuned_recall, 4))
print("Macro F1-score :", round(tuned_f1, 4))
```

```python
# STEP 17: CONFUSION MATRIX FOR TUNED MODEL

from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Predict NDVI using tuned model
tuned_predicted_ndvi = tuned_model.predict(X_test)

# Convert NDVI values to vegetation health classes
def classify_ndvi(ndvi):
    if ndvi < 0.2:
        return "Stressed"
    elif ndvi < 0.5:
        return "Moderate"
    else:
        return "Healthy"

# Convert actual and predicted NDVI into health classes
actual_health = [classify_ndvi(value) for value in y_test]
predicted_health = [classify_ndvi(value) for value in tuned_predicted_ndvi]

# Create confusion matrix
cm_tuned = confusion_matrix(
    actual_health,
    predicted_health,
    labels=["Stressed", "Moderate", "Healthy"]
)

# Plot confusion matrix
plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_tuned,
    annot=True,
    fmt="d",
    xticklabels=["Stressed", "Moderate", "Healthy"],
    yticklabels=["Stressed", "Moderate", "Healthy"]
)

plt.xlabel("Predicted Health")
plt.ylabel("Actual Health")
plt.title("Tuned Model - Vegetation Health Confusion Matrix")
plt.show()
```

```python
# STEP 18: FEATURE IMPORTANCE

import pandas as pd
import matplotlib.pyplot as plt

# Get feature importance from tuned Random Forest
importance = tuned_model.feature_importances_

# Create table
feature_importance = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": importance
})

# Sort from highest to lowest
feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("FEATURE IMPORTANCE")
print("==================")
display(feature_importance)

# Plot
plt.figure(figsize=(8, 5))

plt.bar(
    feature_importance["Feature"],
    feature_importance["Importance"]
)

plt.xlabel("Input Features")
plt.ylabel("Importance")
plt.title("Feature Importance - Tuned Random Forest")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
